```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


# 1. Load Dataset
df = pd.read_csv("laptop_price.csv", encoding="latin1")
 m
print(df.head())
print(df.columns)


# 2. Convert RAM into numbers
df["Ram"] = df["Ram"].str.replace("GB", "").astype(float)


# 3. Select RAM and Price
X = df[["Ram"]]
y = df["Price_euros"]


# 4. Train Regression Model
model = LinearRegression()
model.fit(X, y)


# 5. Predict Price for 16GB RAM
predicted_price = model.predict([[16]])

print("Predicted price for 16GB RAM:",
      round(predicted_price[0], 2), "Euros")


# 6. Check R² Score
print("R² Score:", model.score(X, y))


# 7. Plot
plt.scatter(X, y)
plt.plot(X, model.predict(X))

plt.xlabel("RAM (GB)")
plt.ylabel("Price (Euros)")
plt.title("Laptop Price Prediction using RAM")

plt.show()
```
